```python
from pymongo import MongoClient
from pymongo.errors import DuplicateKeyError, BulkWriteError

# Local Connection
client = MongoClient("mongodb://localhost:27017/")

# Cloud Connection (Atlas)
# uri = "mongodb+srv://<user>:<password>@cluster.mongodb.net/"
# client = MongoClient(uri)

# Access a specific database
db = client["db_02_10"]

# Access a collection (similar to a table in SQL)
users = db["users"]

users.delete_many({})  # Clear the collection before inserting new data

# create a unique index for name field
users.create_index("name", unique=True)
```

```python
# Insert one
user_data = {"name": "Alice", "age": 30, "tags": ["python", "dev"]}

try:
    result = users.insert_one(user_data)
    print(f"Inserted ID: {result.inserted_id}")
except DuplicateKeyError as e:
    print(f"Duplicate key: {e}")

try:
    # Insert many
    user_list = [
        {"name": "Bob", "age": 25},
        {"name": "Charlie", "age": 35}
    ]
    users.insert_many(user_list)
except BulkWriteError as e:
    print(f"Duplicate key: {e}")
```

```python
# Find one by attribute
alice = users.find_one({"name": "Alice"})

# Find many with a filter (e.g., age > 25)
# $gt = Greater Than
for user in users.find({"age": {"$gt": 25}}):
    print(user["name"])
```
