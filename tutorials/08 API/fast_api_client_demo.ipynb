```python
import requests as rq
```

## GET

```python
res = rq.get("http://localhost:8000")
print(res.status_code)
print(res.json())
```

```python
res = rq.get("http://localhost:8000/items/")
print(res.status_code)
print(res.json())
```

```python
res = rq.get("http://localhost:8000/items/1")
print(res.status_code)
print(res.json())
```

## POST

```python
item = {
    "name": "Item 1",
    "description": "This is item 1",
    "price": 10.0
}

res = rq.post("http://localhost:8000/items/", json=item)
print(res.status_code)

item_dict = res.json()
item_id = item_dict['id']
print(item_id)
```

```python
res = rq.get(f"http://localhost:8000/items/{item_id}")
print(res.status_code)
print(res.json())
```

```python
res = rq.get(f"http://localhost:8000/items/")
print(res.status_code)
print(res.json())
```
