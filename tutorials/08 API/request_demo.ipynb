```python
import requests
```

## GET request

```python
# demonostrate a GET request
response = requests.get("https://jsonplaceholder.typicode.com/posts/1")

print(response.status_code) 
print(response.json())
```

```python
res_dict = response.json()
print(res_dict["title"])
```

## POST

```python
# demonstrate POST request
post_data = {   
    "title": "foo",
    "body": "bar",
    "userId": 1
}
response = requests.post("https://jsonplaceholder.typicode.com/posts", json=post_data)
print(response.status_code)
print(response.json())
```

## not found

```python
# demonstrate other status codes
response = requests.get("https://jsonplaceholder.typicode.com/posts/123456789")
print(response.status_code)
```
